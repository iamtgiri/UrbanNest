```python
# Convert Price
def parse_price(x):
    if pd.isna(x): return None
    x = x.replace(',', '').strip().lower()
    if 'lac' in x:
        return float(x.replace('lac', '').strip()) * 1e5
    elif 'cr' in x:
        return float(x.replace('cr', '').strip()) * 1e7
    else:
        return pd.to_numeric(x, errors='coerce')

df['Price'] = df['Price'].apply(parse_price)

# Convert Area
df['Area'] = df['Area'].str.replace(' sqft', '').str.replace(',', '').astype(float)

# Convert Price per Sqft
df['Price per Sqft'] = df['Price per Sqft'].str.replace('/sqft', '').str.replace(',', '').astype(float)
```

```python
import numpy as np

def parse_balcony(value):
    if isinstance(value, str):
        if 'No' in value:
            return 0
        elif '3+' in value:
            return 3
        elif value.strip().isdigit():  # Handles '0'
            return int(value.strip())
        else:
            # Extract the first number from string like "2 Balconies"
            num = ''.join(filter(str.isdigit, value))
            return int(num) if num else np.nan
    elif isinstance(value, (int, float)):
        return int(value)
    return np.nan

df['Balcony'] = df['Balcony'].apply(parse_balcony).astype('Int64')
```

```python
from datetime import datetime

def convert_property_age(value):
    mapping = {
        '0 to 1 Year Old': 0.5,
        '1 to 5 Year Old': 3,
        '5 to 10 Year Old': 7.5,
        '10+ Year Old': 12.5,
        'Within 3 months': -0.25,
        'Within 6 months': -0.5,
        'Under Construction': -1.0,
        'undefined': None
    }

    if value in mapping:
        return mapping[value]
    
    elif isinstance(value, str) and value[:3] in ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                                                  'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']:
        try:
            # Assume current date is July 2025
            today = datetime(2025, 7, 1)
            possession_date = datetime.strptime(value + ' 1', '%b %Y %d')
            delta = (possession_date - today).days / 365
            return round(-delta, 2)  # Negative because it's in future
        except:
            return None
    else:
        return None

df['Property_Age_Years'] = df['Property Age/Possession'].apply(convert_property_age)
```

```python
def classify_furnishing(items):
    if not items or items == '[]':
        return 'Unfurnished'
    count = len(eval(items)) if isinstance(items, str) else len(items)
    if count <= 5:
        return 'Semi-Furnished'
    return 'Fully-Furnished'

df['Furnishing_Level'] = df['Furnishing'].apply(classify_furnishing)
```

```python
df["Furnishing_Level_Code"] = df["Furnishing_Level"].map({
    "Unfurnished": 0,
    "Semi-Furnished": 1,
    "Fully-Furnished": 2
})
```

```python
import ast
from collections import Counter

# Safely parse the facilities into lists without modifying df
parsed_facilities = df['Facilities'].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith('[') else []
)

# Flatten the list and count frequencies
all_facilities = [facility for sublist in parsed_facilities for facility in sublist]
facility_counts = Counter(all_facilities)

# Display results
print(f"Total distinct facilities: {len(facility_counts)}")
print("\nTop 10 most common facilities:")
for facility, count in facility_counts.most_common(31):
    print(f"{facility}: {count}")
```

```python
import ast

# Ensure Facilities is parsed correctly
parsed_facilities = df["Facilities"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[") else [])

# Define your premium facilities
premium_facilities = [
    'Swimming Pool', 'Club house / Community Center', 'Security Personnel',
    'Power Back-up', 'High Ceiling Height', 'Spacious Interiors',
    'Water softening plant', 'Low Density Society', 'Shopping Centre',
    'Private Garden / Terrace', 'Internet/wi-fi connectivity',
    'Centrally Air Conditioned'
]

# Count how many premium facilities are present per row
df["Facility_Premium"] = parsed_facilities.apply(
    lambda x: sum(fac in premium_facilities for fac in x)
)
```

```python
import ast

# Safely convert stringified lists to actual Python lists
parsed_facilities = df["Facilities"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[") else [])


from collections import Counter

facility_counts = Counter()
parsed_facilities.apply(facility_counts.update)

# View top 10
top_10_facilities = [f[0] for f in facility_counts.most_common(10)]


for fac in top_10_facilities:
    df[f'Facility_{fac}'] = parsed_facilities.apply(lambda x: 1 if fac in x else 0)
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
import time
```

```python


# Example: your list of unique locations
location_list = df['Location'].unique()

# Initialize geolocator
geolocator = Nominatim(user_agent="geoapiExercise")

# Define a function to handle timeouts and delays
def get_coordinates(location):
    try:
        time.sleep(1)  # Respect Nominatim's usage policy (1 request/sec)
        loc = geolocator.geocode(location + ", West Bengal, India")  # Add context
        if loc:
            return pd.Series([loc.latitude, loc.longitude])
        else:
            return pd.Series([None, None])
    except GeocoderTimedOut:
        return get_coordinates(location)

# Apply function
geo_df = pd.DataFrame(location_list, columns=['Location'])
geo_df[['Latitude', 'Longitude']] = geo_df['Location'].apply(get_coordinates)

# Preview
print(geo_df.head())
```
